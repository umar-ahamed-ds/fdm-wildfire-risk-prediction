```python
import pandas as pd

df = pd.read_csv(
    "../../data/processed/wildfire_processed.csv"
)

df.shape
```

```python
X = df.drop(
    columns=["wildfire"]
)

y = df["wildfire"]
```

```python
print(X.shape)
print(y.shape)
```

```python
train_mask = X["year"] <= 2020


validation_mask = (
    (X["year"] >= 2021) &
    (X["year"] <= 2022)
)


test_mask = X["year"] >= 2023
```

```python
train = X[train_mask].copy()

train["wildfire"] = y[train_mask]
```

```python
validation = X[validation_mask].copy()

validation["wildfire"] = y[validation_mask]
```

```python
test = X[test_mask].copy()

test["wildfire"] = y[test_mask]
```

```python
print("Train:", train.shape)
print("Validation:", validation.shape)
print("Test:", test.shape)
```

```python
print(
    "Train years:",
    train["year"].min(),
    "-",
    train["year"].max()
)

print(
    "Validation years:",
    validation["year"].min(),
    "-",
    validation["year"].max()
)

print(
    "Test years:",
    test["year"].min(),
    "-",
    test["year"].max()
)
```

```python
train.to_csv(
    "../../data/processed/train.csv",
    index=False
)


validation.to_csv(
    "../../data/processed/validation.csv",
    index=False
)


test.to_csv(
    "../../data/processed/test.csv",
    index=False
)
```

```python
train_years = set(train["year"].unique())
validation_years = set(validation["year"].unique())
test_years = set(test["year"].unique())


print("Train & Validation:")
print(train_years.intersection(validation_years))


print("\nTrain & Test:")
print(train_years.intersection(test_years))


print("\nValidation & Test:")
print(validation_years.intersection(test_years))
```

```python
print("Train distribution")
print(
    train["wildfire"].value_counts(normalize=True)
)


print("\nValidation distribution")
print(
    validation["wildfire"].value_counts(normalize=True)
)


print("\nTest distribution")
print(
    test["wildfire"].value_counts(normalize=True)
)
```
